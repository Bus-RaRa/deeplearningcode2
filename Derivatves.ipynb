```python
import numpy as np
import matplotlib.pyplot as plt
#smpy = sembolic math in Python
import sympy as sym
import sympy.plotting.plot as symplot
```

```python
#create symbolic variables in sympy
x = sym.symbols('x')

#create a function
fx = 2*x**2

#computate its derivative
df = sym.diff(fx,x)

#print them
print(fx)
print(df)
```

```python
symplot(fx,(x,-4,4),title='the function')
plt.show()

symplot(df,(x,-4,4),title='the derivative')
plt.show()
```

```python
#repeat with relu and sigmoid

#create symbolic functions
relu = sym.Max(0,x)
sigmoid = 1/(1+sym.exp(-x))

#compute their derivatives
df_relu = sym.diff(relu,x)
df_sigmoid = sym.diff(sigmoid,x)

p = symplot(relu,(x,-4,4),label='relu',show=False,line_color='b')
p.extend(symplot(sigmoid,(x,-4,4),label='sigmoid',show=False,line_color='r'))
p.legend=True
p.title='the functions'
p.show()

#graph their derivatives
p = symplot(df_relu,(x,-4,4),label='relu',show=False,line_color='b')
p.extend(symplot(df_sigmoid,(x,-4,4),label='sigmoid',show=False,line_color='r'))
p.legend=True
p.title='the derivatives'
p.show()
```

#Chain Rule

```python
from IPython.display import display
```

```python
#create symbolic variables in sympy
x = sym.symbols('x')

#create a function
fx = 2*x**2
gx = 4*x**3 - 3*x**4

#computate its derivative
df = sym.diff(fx,x)
dg = sym.diff(gx,x)

#appyly the product rule "manually"
manual= df*gx + fx*dg
thewrongway = df*dg

#via sympy
viasympy = sym.diff(fx*gx)


#print them
print('the functions:')
display(fx)
display(gx)
print('the derivatives:')
display(df)
display(dg)
print('the product rule:')
display(manual)
display(thewrongway)
display(viasympy)
```

```python
#repeat with chain rule
fx = (x**2 + 4*x**3)**5

print('the function')
display(fx)
print(' ')

df = sym.diff(fx)
print('the derivative')
display(df)
```
