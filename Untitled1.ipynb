```python
import numpy as np
import torch
import torch.nn as nn
import torch.nn.functional as F
import matplotlib.pyplot as plt
```

```python
#manually in numpy
#the list of numbers
z = [1,2,3]

#compute the softmax result
num = np.exp(z)
den = np.sum(np.exp(z))
softmax = num/den
print(softmax)
print(np.sum(softmax))
```

```python
#pytorch
softfun = nn.Softmax(dim=0)
z_tensor = softfun(torch.Tensor(z))
print(z_tensor)
```

```python
x = [0.25,0.75]
H = 0
for p in x:
  H -= p*np.log(p)
print(H)
```

```python
#cross entropy
```

```python
p = [1,0]
q = [0.25, 0.75]
H = 0
for i in range(len(p)):
  H -= p[i]*np.log(q[i])
print(H)
```

```python
#simplification
H = -np.log(q[0])
```
