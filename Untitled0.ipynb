```python
import numpy as np
import torch
```

```python
#using numpy
nv1 = np.array([1,2,3,4])
nv2 = np.array([0,1,0,-1])
#dot product via function
print(np.dot(nv1,nv2))
#dot product via scalar multiplication and sum
print(np.sum(nv1*nv2))
```

```python
#USING PYTORCH
```

```python
#create a vector
tv1 = torch.tensor([1,2,3,4])
tv2 = torch.tensor([0,1,0,-1])
#dot product via function
print(torch.dot(tv1,tv2))
#dot product via scalar multiplication and sum
print(torch.sum(tv1*tv2))
```

```python
#MATRİX MULTİPLİCATİON
#create some random matrices
A = np.random.randn(3,4)
B = np.random.randn(4,5)
C = np.random.randn(3,7)

#try some multiplications
np.matmul(A,B)
print(np.round(A@B,2)), print("")
print(np.round(C.T@A,2)), print("")
```

```python
#create some random matrics
A = torch.randn(3,4)
B = torch.randn(4,5)
C1 = np.random.randn(4,7)
C2 = torch.tensor(C1,dtype=torch.float)

#try some multiplications
print(np.round(A@B,2)), print("")
print(np.round(A@B.T,2)), print("")
print(np.round(A@C1,2)), print("")
print(np.round(A@C2,2)), print("")
```
